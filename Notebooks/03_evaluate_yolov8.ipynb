# YOLOv8 Evaluation (RDD2022)

This notebook evaluates a trained YOLOv8 model on the labeled test split and runs a few sample predictions.

## 1) Install and import

```python
!pip -q install ultralytics
```

```python
from pathlib import Path
from ultralytics import YOLO

DATA_YAML = Path("/kaggle/working/rdd2022_yolo/data.yaml")
WEIGHTS = Path("/kaggle/working/yolo_runs/rdd2022_yolov8s/weights/best.pt")
DATA_YAML, WEIGHTS
```

## 2) Evaluate on labeled test split

```python
model = YOLO(str(WEIGHTS))
metrics = model.val(data=str(DATA_YAML), split="test", imgsz=640, batch=16)
metrics
```

## 3) Run predictions on a few test images

```python
test_img_dir = Path("/kaggle/working/rdd2022_yolo/images/test")
sample_imgs = list(test_img_dir.glob("*.jpg"))[:5]
pred = model.predict(source=[str(p) for p in sample_imgs], imgsz=640, conf=0.25, save=True)
pred
```

## 4) Optional: run inference on original dataset test images (if unlabeled)
Use this if you want to see predictions on the official test images.

```python
orig_test_imgs = Path("/kaggle/input/rdd2022/rdd-2022/Japan/test/images")
if orig_test_imgs.exists():
    img_list = list(orig_test_imgs.glob("*.jpg"))[:5]
    model.predict(source=[str(p) for p in img_list], imgsz=640, conf=0.25, save=True)
else:
    print("Original test images not found for this country path.")
```
